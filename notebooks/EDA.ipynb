```python

stop_rows = []

for route_id, route in route_data.items():
    for stop_id, stop in route["stops"].items():
        stop_rows.append({
            "route_id": route_id,
            "stop_id": stop_id,
            "station_code": route["station_code"],
            **stop
        })

stops_df = pd.DataFrame(stop_rows)

stops_df.head()
```

```python
outliers = pd.Series(False, index=stops_df.index)

for col in ["lat", "lng"]:
    q1 = stops_df.groupby("station_code")[col].transform(
        lambda x: x.quantile(0.25)
    )
    q3 = stops_df.groupby("station_code")[col].transform(
        lambda x: x.quantile(0.75)
    )

    iqr = q3 - q1

    outliers |= (
        (stops_df[col] < q1 - 3 * iqr) |
        (stops_df[col] > q3 + 3 * iqr)
    )

suspicious_stops = stops_df[outliers]

suspicious_stops.groupby("station_code").size()
```

```python
stops_df["suspicious"] = outliers

stops_df.groupby("station_code")["suspicious"].agg(["sum", "count"])
```

```python
station = stops_df[
    stops_df["station_code"] == "DAU1"
][["lat", "lng"]].dropna().copy()
```

```python
from pyproj import Transformer

# Determine the appropriate UTM zone for this station
lat = station["lat"].mean()
lng = station["lng"].mean()

zone = int((lng + 180) // 6) + 1
epsg = (32600 if lat >= 0 else 32700) + zone

transformer = Transformer.from_crs(
    "EPSG:4326",
    f"EPSG:{epsg}",
    always_xy=True
)

station["x_m"], station["y_m"] = transformer.transform(
    station["lng"].to_numpy(),
    station["lat"].to_numpy()
)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import pandas as pd

X = station[["x_m", "y_m"]]

inertias = []
silhouette_scores = []

k_values = range(2, min(15, len(X) - 1) + 1)

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    inertias.append(model.inertia_)

    silhouette_scores.append(
        silhouette_score(
            X,
            labels,
            sample_size=min(1500, len(X)),
            random_state=42
        )
    )
```

```python
plt.plot(k_values, inertias, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
plt.plot(k_values, silhouette_scores, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Analysis")
plt.show()
```
